```python
from pathlib import Path
import numpy as np
import pandas as pd
from flycraft.utils.geometry_utils import angle_of_2_velocity
PROJECT_ROOT_DIR = Path().absolute().parent.parent.parent
PROJECT_ROOT_DIR
```

```python
def process_one_file(file_path: Path):
    df = pd.read_csv(file_path)
    df["v_error"] = np.abs(df["v"] - df["achieved v"])
    #df["angle_error"] = df.apply(lambda row: np.abs(angle_of_2_velocity(row["v"], row["mu"], row["chi"], row["achieved v"], row["achieved mu"], row["achieved chi"])), axis=1)
    success_rate = len(df[df["termination"] == "reach target"])/len(df)
    df_timeout = df[(df["termination"] == "timeout")]
    len_of_timeout = len(df_timeout)
    average_angle_error =  sum(df_timeout["error"]) / len_of_timeout
    average_v_error = sum(df_timeout["v_error"]) / len_of_timeout
    return df_timeout,len_of_timeout,average_angle_error,average_v_error,success_rate
```

```python
files = [
    "logs/cache/LadderRL/eval_error/b_2_one_stage_skip_5_5e6steps_hard_1.csv", 
    "logs/cache/LadderRL/eval_error/b_2_one_stage_skip_5_5e6steps_hard_2.csv", 
    "logs/cache/LadderRL/eval_error/b_2_one_stage_skip_5_5e6steps_hard_3.csv", 
    "logs/cache/LadderRL/eval_error/b_2_one_stage_skip_5_5e6steps_hard_4.csv", 
    "logs/cache/LadderRL/eval_error/b_2_one_stage_skip_5_5e6steps_hard_5.csv", 
]

outcome = []
for index,file in enumerate(files):
    df_timeout,lens,angle_error,v_error,success_rate = process_one_file(PROJECT_ROOT_DIR / file )
    current_outcome = {
        "seed": index+1,
        "len_of_timeout": lens,
        "angle": angle_error,
        "v":v_error,
        "success_rate":success_rate,
    }
    outcome.append(current_outcome)
print(outcome)

# 计算平均值
avg_length = sum(item['len_of_timeout'] for item in outcome) / len(outcome)
avg_angle = sum(item['angle'] for item in outcome) / len(outcome)
avg_velocity = sum(item['v'] for item in outcome) / len(outcome)
avg_successrate = sum(item['success_rate'] for item in outcome) / len(outcome)
print(f"Average length: {avg_length:.2f}")
print(f"Average angle: {avg_angle:.2f}")
print(f"Average velocity: {avg_velocity:.2f}")
print(f"Average successrate: {avg_successrate:.2f}")
```
